```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from catboost import CatBoostClassifier
from sklearn.utils import class_weight as sk_class_weight
from sklearn.preprocessing import LabelEncoder 
import numpy as np 

# ==============================================================================
# 1. DATA LOADING AND CLEANING 
# ==============================================================================

# Load data (assuming file is in the current directory)
data_native = pd.read_csv("mentalhealthdataset.csv")
data_native = data_native.drop(columns=['Timestamp'])
data_native['self_employed'] = data_native['self_employed'].fillna('No')
data_native = data_native.drop_duplicates()

# ==============================================================================
# 2. TARGET COLUMN CREATION (NON-DETERMINISTIC LOGIC)
# ==============================================================================

# Temporary numeric columns for defining the risk logic
data_native['Mood_Swings_Numeric'] = data_native['Mood_Swings'].map({'Low': 0, 'Medium': 1, 'High': 2})
data_native['Mental_Health_History_Numeric'] = data_native['Mental_Health_History'].map({'No': 0, 'Maybe': 1, 'Yes': 2})
data_native['treatment_Numeric'] = data_native['treatment'].map({'No': 0, 'Yes': 1}) 

def define_risk_native(row):
    if (row['treatment_Numeric'] == 1 or
        (row['Mental_Health_History_Numeric'] == 2 and row['Mood_Swings_Numeric'] == 2)):
        return 'High'
    if (row['treatment_Numeric'] == 0 and
        row['Mental_Health_History_Numeric'] == 0 and
        row['Mood_Swings_Numeric'] == 0):
        return 'Low'
    return 'Medium'

data_native['Risk_Level'] = data_native.apply(define_risk_native, axis=1)

# Drop the temporary encoded columns
data_native = data_native.drop(columns=['Mood_Swings_Numeric', 'Mental_Health_History_Numeric', 'treatment_Numeric'])


# ==============================================================================
# 3. FEATURE DEFINITION AND ENCODING FOR CATBOOST (FIXED)
# ==============================================================================

# 3.1 ORDAINALLY ENCODE ALL SIMPLE/BINARY CATEGORICAL FEATURES TO INTEGERS
# This solves the CatBoostError by converting all non-nominal strings to numbers.
simple_cat_cols = ['Gender', 'family_history', 'Coping_Struggles', 'Work_Interest', 'Social_Weakness', 
                   'mental_health_interview', 'care_options', 'Growing_Stress', 'Changes_Habits', 'self_employed'] # Added self_employed

for col in simple_cat_cols:
    data_native[col] = LabelEncoder().fit_transform(data_native[col])


# Ordinal Encoding for Days_Indoors (retaining rank)
ordinal_map = {'Go out Every day': 0, '1-14 days': 1, '15-30 days': 2, '31-60 days': 3, 'More than 2 months': 4}
data_native['Days_Indoors'] = data_native['Days_Indoors'].map(ordinal_map)


# CRITICAL: Exclude the original strings used in the target logic.
# These columns are still present in their original string form and must be excluded.
X_native = data_native.drop(columns=['Risk_Level', 'treatment', 'Mental_Health_History', 'Mood_Swings'])
y_native = data_native['Risk_Level']

# 3.2 IDENTIFY NATIVE CATBOOST FEATURES (Only Country and Occupation remain as 'object' type strings)
CAT_FEATURES = X_native.select_dtypes(include=['object']).columns.tolist()


# ==============================================================================
# 4. DATA SPLIT, CLASS WEIGHTING, AND CATBOOST TRAINING
# ==============================================================================

# Split data (Stratified split on the STRING target)
X_train_nat, X_test_nat, y_train_nat, y_test_nat = train_test_split(
    X_native, y_native, test_size=0.3, random_state=42, stratify=y_native
)

# 4.1 Algorithm-Level Imbalance Handling (Using Native Class Weights)
y_labels = y_train_nat.unique()
class_weights = sk_class_weight.compute_class_weight(
    'balanced', classes=y_labels, y=y_train_nat
)
weights_dict = dict(zip(y_labels, class_weights))


# Map target values to numeric for CatBoost (0, 1, 2)
label_map = {'Low': 0, 'Medium': 1, 'High': 2}
y_train_catboost = y_train_nat.map(label_map) 


# 4.2 Initialize CatBoost Model (Faster Benchmark Settings)
cat_model_native = CatBoostClassifier(
    loss_function='MultiClass',
    iterations=100,             
    depth=8,                   
    learning_rate=0.1,        
    random_state=42,
    verbose=0, 
    cat_features=CAT_FEATURES,  # CatBoost handles Country/Occupation strings natively
    class_weights=weights_dict, 
)

print("\nFitting Pure CatBoost Model with Native Encoding and Class Weights...")
cat_model_native.fit(X_train_nat, y_train_catboost, cat_features=CAT_FEATURES)


# 4.3 Predict and Evaluate
y_pred_cat_native_int = cat_model_native.predict(X_test_nat)
y_pred_cat_native_int = y_pred_cat_native_int.flatten() 

# Convert predictions back to string labels for reporting
reverse_label_map = {0: 'Low', 1: 'Medium', 2: 'High'}
y_pred_cat_native = pd.Series(y_pred_cat_native_int).map(reverse_label_map)


print("\n--- CatBoost Classification Report (Native Encoding + Class Weights) ---\n")
print(classification_report(y_test_nat, y_pred_cat_native))
print("CatBoost Confusion Matrix:\n", confusion_weight(y_test_nat, y_pred_cat_native))
```
